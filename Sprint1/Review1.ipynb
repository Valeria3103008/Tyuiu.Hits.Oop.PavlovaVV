```
using System.Collections.Generic;

Product myProduct = new Product("Ноутбук", 65000m, 10, "Электроника", "Игровой ноутбук с хорошей видеокартой");
Seller mySeller = new Seller("Валерия", "EMP-001", 50000m, "pavlova@store.com");
Store myStore = new Store("ТехноМир", "ул. Мельникайте, 70", "09:00 - 21:00");

myStore.AddSeller(mySeller);
mySeller.AddProduct(myProduct);

Console.WriteLine(myStore.GetStoreInfo());
Console.WriteLine(mySeller.GetSellerInfo());
Console.WriteLine($"Общая стоимость на складе: {myProduct.GetTotalPrice()} руб.");

Console.WriteLine();
mySeller.SellProduct(myProduct, 2);

Console.WriteLine("\n--- Все товары магазина ---");
myStore.ListProducts();



public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name { get { return name; } set { name = value; } }
    public decimal Price { get { return price; } set { price = value >= 0 ? value : 0; } }
    public int Quantity { get { return quantity; } set { quantity = value >= 0 ? value : 0; } }
    public string Category { get { return category; } set { category = value; } }
    public string Description { get { return description; } set { description = value; } }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        Name = name;
        Price = price;
        Quantity = quantity;
        Category = category;
        Description = description;
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity += amount;
    }

    public string GetProductInfo()
    {
        return $"Товар: {Name} (Категория: {Category}). Описание: {Description}. Цена: {Price} руб. В наличии: {Quantity} шт.";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    
    public List<Product> SellerProducts { get; private set; } = new List<Product>();

    public string Name { get { return name; } set { name = value; } }
    public string EmployeeId { get { return employeeId; } set { employeeId = value; } }
    public decimal Salary { get { return salary; } set { salary = value >= 0 ? value : 0; } }
    public string ContactInfo { get { return contactInfo; } set { contactInfo = value; } }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        Name = name;
        EmployeeId = employeeId;
        Salary = salary;
        ContactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        SellerProducts.Add(product);
    }

    public void SellProduct(Product product, int sellQuantity)
    {
        if (product.Quantity >= sellQuantity)
        {
            product.UpdateQuantity(-sellQuantity);
            Console.WriteLine($"Успешно продано {sellQuantity} шт. товара «{product.Name}».");
        }
        else
        {
            Console.WriteLine($"Ошибка: недостаточно товара «{product.Name}» на складе.");
        }
    }

    public string GetSellerInfo()
    {
        return $"Продавец: {Name} (Табельный номер: {EmployeeId}). Контакты: {ContactInfo}";
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    
    private List<Seller> Sellers { get; set; } = new List<Seller>();

    public string StoreName { get { return storeName; } set { storeName = value; } }
    public string Location { get { return location; } set { location = value; } }

    public Store(string storeName, string location, string storeHours)
    {
        StoreName = storeName;
        Location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        Sellers.Add(seller);
    }

    public void ListProducts()
    {
        foreach (Seller seller in Sellers)
        {
            foreach (Product product in seller.SellerProducts)
            {
                Console.WriteLine(product.GetProductInfo());
            }
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин «{StoreName}» по адресу: {Location}. Часы работы: {storeHours}";
    }
}
```
